# 1. Introduction to Probability

### Definition

Probability measures how likely an event is to happen.

Its value ranges from 0 to 1.

0 means the event is impossible, while 1 means the event is certain.

### Real-life Example

When a fair coin is tossed, there are two possible outcomes: Head and Tail.

The probability of getting Head is 1 out of 2.

### Formula

$$
P(A)=\frac{\text{Number of favorable outcomes}}{\text{Total number of possible outcomes}}
$$

Where:

- $P(A)$ = probability of event A
- $A$ = event we are interested in

### Numerical Example

For a fair coin:

$$
P(\text{Head})=\frac{1}{2}=0.5
$$

Therefore, the probability of getting Head is 0.5 or 50%.

### Python Implementation


```python
favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes

print("Probability:", probability)
```

### AI/ML Example

Machine learning classification models often produce probability estimates for different classes.


```python
spam_probability = 0.85

if spam_probability > 0.5:
    prediction = "Spam"
else:
    prediction = "Not Spam"

print("Prediction:", prediction)
```

### Explanation

The code uses a predicted probability of 0.85 for the Spam class.

Because the probability is greater than the chosen threshold of 0.5, the example classifies the message as Spam.

Probabilities are useful in classification because they represent the model's estimated likelihood for each class and can be converted into class predictions using a decision threshold.


# 2. Events

### Definition

An event is a specific outcome or collection of outcomes from a random experiment.

### Real-life Example

When rolling a dice, getting an even number is an event.

The event contains the outcomes:

2, 4, 6

### Formula

If event A represents getting an even number:

$$
A=\{2,4,6\}
$$

The probability is:

$$
P(A)=\frac{3}{6}=\frac{1}{2}
$$

Where:

- $A$ = event containing the desired outcomes
- $P(A)$ = probability that event A occurs

### Numerical Example

A fair dice has six outcomes.

Three outcomes are even.

$$
P(A)=\frac{3}{6}=0.5
$$

### Python Implementation


```python
dice = [1, 2, 3, 4, 5, 6]

event = [x for x in dice if x % 2 == 0]

probability = len(event) / len(dice)

print("Event:", event)
print("Probability:", probability)
```

### AI/ML Example

In a classification problem, an event can represent a particular class outcome.


```python
prediction_probabilities = {
    "Cat": 0.20,
    "Dog": 0.70,
    "Bird": 0.10
}

event_probability = prediction_probabilities["Dog"]

print("Probability of Dog:", event_probability)
```

### Explanation

Here, the event is that the image belongs to the Dog class.

The model assigns a probability of 0.70 to this event.

Thinking in terms of events helps us interpret classification probabilities and reason about possible outcomes produced by a model.


# 3. Sample Space

### Definition

Sample space is the set of all possible outcomes of a random experiment.

It is usually represented by $S$ or $\Omega$.

### Real-life Example

When tossing one coin, the possible outcomes are Head and Tail.

Therefore, the sample space contains both outcomes.

### Formula

$$
S=\{H,T\}
$$

Where:

- $S$ = sample space
- $H$ = Head
- $T$ = Tail

### Numerical Example

For a six-sided dice:

$$
S=\{1,2,3,4,5,6\}
$$

The total number of possible outcomes is:

$$
|S|=6
$$

Where:

- $|S|$ = number of outcomes in the sample space

### Python Implementation


```python
sample_space = [1, 2, 3, 4, 5, 6]

print("Sample Space:", sample_space)
print("Number of Outcomes:", len(sample_space))
```

### AI/ML Example

For a three-class classification problem, the possible classes form the set of possible output outcomes.


```python
classes = ["Cat", "Dog", "Bird"]

print("Possible Classes:", classes)
```

### Explanation

The code defines all classes that the classification model can predict.

Understanding the possible outcome space is important because probability predictions are assigned across these possible classes.


# 4. Independent Events

### Definition

Independent events are events where the occurrence of one event does not change the probability of the other event.

### Real-life Example

Tossing a coin twice gives independent events.

Getting Head on the first toss does not change the probability of getting Head on the second toss.

### Formula

$$
P(A \cap B)=P(A)\times P(B)
$$

Where:

- $P(A)$ = probability of event A
- $P(B)$ = probability of event B
- $A \cap B$ = both A and B occurring

### Numerical Example

Probability of Head on the first toss:

$$
P(A)=\frac{1}{2}
$$

Probability of Head on the second toss:

$$
P(B)=\frac{1}{2}
$$

Therefore:

$$
P(A \cap B)=\frac{1}{2}\times\frac{1}{2}
$$

$$
P(A \cap B)=\frac{1}{4}=0.25
$$

### Python Implementation


```python
p_a = 0.5
p_b = 0.5

p_both = p_a * p_b

print("Probability of Both:", p_both)
```

### AI/ML Example

Independent probability assumptions are used in models such as Naive Bayes.


```python
p_word1 = 0.6
p_word2 = 0.4

combined_probability = p_word1 * p_word2

print("Combined Probability:", combined_probability)
```

### Explanation

The example multiplies two probabilities under an independence assumption.

Naive Bayes simplifies probability calculations by assuming that features are conditionally independent given the class.

This assumption makes the model computationally simple even when a dataset contains many features.


# 5. Dependent Events

### Definition

Dependent events are events where the occurrence of one event changes the probability of another event.

### Real-life Example

Suppose two cards are drawn from a deck without replacement.

After the first card is removed, the number of cards remaining changes.

Therefore, the probability of the second event depends on what happened in the first event.

### Formula

$$
P(A \cap B)=P(A)\times P(B|A)
$$

Where:

- $P(A)$ = probability of event A
- $P(B|A)$ = probability of event B given that A has occurred
- $A \cap B$ = both events occurring

### Numerical Example

Probability of drawing an Ace first:

$$
P(A)=\frac{4}{52}
$$

After drawing one Ace, probability of drawing another Ace:

$$
P(B|A)=\frac{3}{51}
$$

Therefore:

$$
P(A \cap B)=\frac{4}{52}\times\frac{3}{51}
$$

$$
P(A \cap B)\approx0.00452
$$

### Python Implementation


```python
p_first_ace = 4 / 52
p_second_ace_given_first = 3 / 51

p_both = p_first_ace * p_second_ace_given_first

print("Probability of Two Aces:", p_both)
```

### AI/ML Example

Dependencies between features can affect how probability models are designed.


```python
p_feature_a = 0.6
p_feature_b_given_a = 0.8

joint_probability = p_feature_a * p_feature_b_given_a

print("Joint Probability:", joint_probability)
```

### Explanation

The code calculates the probability of two related events occurring together.

The probability of the second feature is considered after knowing that the first feature occurred.

Understanding dependence is important because real-world features are often related, and assuming independence when strong dependencies exist can affect a probabilistic model.


# 6. Conditional Probability

### Definition

Conditional probability is the probability of an event occurring when another event is already known to have occurred.

### Real-life Example

Suppose we want to find the probability that a student passed Mathematics given that the student attended extra classes.

We calculate the probability using only students who attended the extra classes.

### Formula

$$
P(A|B)=\frac{P(A \cap B)}{P(B)}
$$

Where:

- $P(A|B)$ = probability of A given B
- $P(A \cap B)$ = probability that both A and B occur
- $P(B)$ = probability of event B

### Numerical Example

Suppose:

$$
P(A \cap B)=0.30
$$

$$
P(B)=0.50
$$

Then:

$$
P(A|B)=\frac{0.30}{0.50}=0.60
$$

Therefore, the conditional probability is 60%.

### Python Implementation


```python
p_a_and_b = 0.30
p_b = 0.50

conditional_probability = p_a_and_b / p_b

print("P(A|B):", conditional_probability)
```

### AI/ML Example

Conditional probabilities are central to probabilistic classification.


```python
p_spam = 0.30
p_word_given_spam = 0.80

joint_probability = p_spam * p_word_given_spam

print("P(Spam and Word):", joint_probability)
```

### Explanation

The example uses the probability of a word appearing given that an email is spam.

Conditional probabilities allow machine learning models to use known evidence when estimating the probability of a class.

This idea is fundamental to models such as Naive Bayes.


# 7. Bayes' Theorem

### Definition

Bayes' Theorem updates the probability of an event after receiving new evidence.

It combines prior knowledge with new information.

### Real-life Example

Suppose a medical test is positive.

Bayes' Theorem can be used to estimate the probability that the person actually has the disease by considering the disease rate and the accuracy of the test.

### Formula

$$
P(A|B)=
\frac{P(B|A)\times P(A)}
{P(B)}
$$

Where:

- $P(A|B)$ = posterior probability, probability of A after observing B
- $P(B|A)$ = likelihood, probability of observing B when A is true
- $P(A)$ = prior probability of A
- $P(B)$ = probability of the evidence B

### Numerical Example

Suppose:

$$
P(A)=0.20
$$

$$
P(B|A)=0.80
$$

$$
P(B)=0.40
$$

Then:

$$
P(A|B)=
\frac{0.80\times0.20}{0.40}
$$

$$
P(A|B)=0.40
$$

### Python Implementation


```python
p_a = 0.20
p_b_given_a = 0.80
p_b = 0.40

p_a_given_b = (p_b_given_a * p_a) / p_b

print("P(A|B):", p_a_given_b)
```

### AI/ML Example

Bayes' Theorem forms the foundation of Naive Bayes classification.


```python
p_spam = 0.30
p_word_given_spam = 0.80
p_word = 0.50

p_spam_given_word = (
    p_word_given_spam * p_spam
) / p_word

print("P(Spam|Word):", p_spam_given_word)
```

### Explanation

The code starts with the prior probability that an email is spam and updates it using evidence that a particular word appeared.

The result represents the probability of Spam after observing the word.

Bayesian reasoning is useful in machine learning because predictions can be updated when new evidence becomes available.


# 8. Probability Rules

### Definition

Probability rules are mathematical rules used to calculate probabilities when events are combined.

Important rules include the complement rule, addition rule and multiplication rule.

### Real-life Example

When rolling a dice, we may want to calculate the probability of getting either an even number or a number greater than 4.

Probability rules help combine these events correctly.

### Formula

Complement Rule:

$$
P(A^c)=1-P(A)
$$

Where:

- $A^c$ = event A does not occur
- $P(A)$ = probability of event A

Addition Rule:

$$
P(A \cup B)
=
P(A)+P(B)-P(A \cap B)
$$

Where:

- $A \cup B$ = A or B or both occur
- $A \cap B$ = both A and B occur

For mutually exclusive events:

$$
P(A \cup B)=P(A)+P(B)
$$

Multiplication Rule:

$$
P(A \cap B)=P(A)\times P(B|A)
$$

For independent events:

$$
P(A \cap B)=P(A)\times P(B)
$$

### Numerical Example

Suppose:

$$
P(A)=0.40
$$

$$
P(B)=0.30
$$

$$
P(A \cap B)=0.10
$$

Using the addition rule:

$$
P(A \cup B)=0.40+0.30-0.10
$$

$$
P(A \cup B)=0.60
$$

### Python Implementation


```python
p_a = 0.40
p_b = 0.30
p_a_and_b = 0.10

p_a_or_b = p_a + p_b - p_a_and_b

print("P(A or B):", p_a_or_b)
```

### AI/ML Example

Probability rules can be used to combine probabilities when analysing possible outcomes.


```python
p_positive = 0.70

p_negative = 1 - p_positive

print("P(Positive):", p_positive)
print("P(Negative):", p_negative)
```

### Explanation

The example uses the complement rule.

If a binary classifier assigns probability 0.70 to the positive class, the complementary probability is 0.30 when the two classes are exhaustive and mutually exclusive.

Probability rules help ensure that probabilistic outputs are combined and interpreted consistently.


# 9. Random Variables

### Definition

A random variable assigns a numerical value to each possible outcome of a random experiment.

Random variables can be discrete or continuous.

### Real-life Example

When rolling a dice, let X represent the number shown on the dice.

Possible values are:

$$
X=\{1,2,3,4,5,6\}
$$

### Formula

A discrete random variable can be described using:

$$
P(X=x)
$$

Where:

- $X$ = random variable
- $x$ = one possible value of the random variable
- $P(X=x)$ = probability that X takes the value x

For all possible values:

$$
\sum_x P(X=x)=1
$$

Where:

- $\sum$ = sum over all possible values of X

### Numerical Example

For a fair dice:

$$
P(X=4)=\frac{1}{6}
$$

Each possible dice value has probability:

$$
\frac{1}{6}
$$

### Python Implementation


```python
import random

X = random.randint(1, 6)

print("Random Variable X:", X)
```

### AI/ML Example

A model prediction can be treated as a random variable with possible class outcomes.


```python
import numpy as np

classes = ["Cat", "Dog", "Bird"]
probabilities = [0.2, 0.7, 0.1]

prediction = np.random.choice(
    classes,
    p=probabilities
)

print("Random Prediction:", prediction)
```

### Explanation

The code defines three possible class outcomes and their probabilities.

A class is randomly selected according to those probabilities.

Random variables are important in probability-based machine learning because uncertain outcomes, predictions and model parameters can be represented mathematically using probability distributions.


# 10. Expected Value

### Definition

Expected value is the long-run average value we expect from a random variable when an experiment is repeated many times.

It is a probability-weighted average of all possible outcomes.

### Real-life Example

Suppose a game gives:

₹100 with probability 0.6  
₹0 with probability 0.4

The expected value tells us the average amount we would expect to receive per game over many repetitions.

### Formula

For a discrete random variable:

$$
E(X)=\sum_i x_iP(x_i)
$$

Where:

- $E(X)$ = expected value of random variable X
- $x_i$ = each individual possible value
- $P(x_i)$ = probability of that value
- $\sum_i$ = sum over all possible values

### Numerical Example

Suppose:

$$
x_1=100,\quad P(x_1)=0.6
$$

$$
x_2=0,\quad P(x_2)=0.4
$$

Then:

$$
E(X)
=
(100\times0.6)+(0\times0.4)
$$

$$
E(X)=60
$$

The expected value is ₹60.

### Python Implementation


```python
values = [100, 0]
probabilities = [0.6, 0.4]

expected_value = sum(
    value * probability
    for value, probability in zip(values, probabilities)
)

print("Expected Value:", expected_value)
```

### AI/ML Example

Expected value can be used when a model predicts several possible numerical outcomes with different probabilities.


```python
outcomes = [10, 20, 30]
probabilities = [0.2, 0.5, 0.3]

expected_prediction = sum(
    outcome * probability
    for outcome, probability in zip(outcomes, probabilities)
)

print("Expected Prediction:", expected_prediction)
```

### Explanation

The code multiplies each possible outcome by its probability and adds the results.

This gives a probability-weighted average prediction.

Expected value is useful in probabilistic machine learning and decision-making because it summarizes uncertain numerical outcomes into one average expected result.
