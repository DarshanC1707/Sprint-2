# Probability Distributions


# 1. Bernoulli Distribution


### Definition
Bernoulli distribution represents a single experiment that has only two possible outcomes: success or failure.

Usually, success is represented by 1 and failure by 0.

### Formula

$$
P(X=x)=p^x(1-p)^{1-x}, \quad x\in\{0,1\}
$$

Where:

- $X$ = Bernoulli random variable
- $x$ = observed outcome, either 0 or 1
- $p$ = probability of success
- $1-p$ = probability of failure

### Where it is used
It is used when an experiment has exactly two possible outcomes, such as yes/no, pass/fail, clicked/not clicked, or fraud/not fraud.

### Real-world Example
Suppose a customer either clicks an advertisement or does not click it.

Click = 1  
No Click = 0

If the probability of clicking is 0.7:

$$
P(X=1)=0.7
$$

$$
P(X=0)=0.3
$$

### AI/ML Example
Bernoulli outcomes are common in binary classification problems.

For example, a model may predict whether an email is spam or not spam.

### Python Visualization


```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.7
data = np.random.binomial(1, p, 1000)

values, counts = np.unique(data, return_counts=True)

plt.bar(values, counts / len(data))
plt.xticks([0, 1], ["Failure", "Success"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.show()
```

### Explanation
The code generates 1000 Bernoulli trials where the probability of success is 0.7.

Each trial produces either 0 or 1. In AI/ML, this type of distribution is useful for modelling binary outcomes such as spam/not spam or fraud/not fraud.


# 2. Binomial Distribution


### Definition
Binomial distribution represents the number of successes in a fixed number of independent Bernoulli trials.

Each trial has the same probability of success.

### Formula

$$
P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}
$$

Where:

- $X$ = number of successes
- $k$ = specific number of successes
- $n$ = total number of independent trials
- $p$ = probability of success in one trial
- $1-p$ = probability of failure
- $\binom{n}{k}$ = number of ways to choose k successes from n trials

### Where it is used
It is used when we repeat the same independent success/failure experiment a fixed number of times.

### Real-world Example
Suppose a coin is tossed 10 times.

We want to calculate the probability of getting exactly 6 Heads.

Here:

$$
n=10,\quad k=6,\quad p=0.5
$$

### AI/ML Example
Binomial distribution can model the number of successful predictions in a fixed number of independent trials when the probability of success is assumed to remain constant.

For example, it can be used in simple probability models of repeated binary outcomes.

### Python Visualization


```python
import numpy as np
import matplotlib.pyplot as plt

n = 10
p = 0.5

data = np.random.binomial(n, p, 1000)

values, counts = np.unique(data, return_counts=True)

plt.bar(values, counts / len(data))
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

### Explanation
The code simulates 1000 experiments. Each experiment contains 10 independent trials with a success probability of 0.5.

The graph shows how frequently different numbers of successes occur. This is useful when analysing repeated binary outcomes under fixed-trial assumptions.


# 3. Poisson Distribution


### Definition
Poisson distribution represents the number of times an event occurs within a fixed interval of time, distance, area, or another exposure when events occur at an approximately constant average rate and independently.

### Formula

$$
P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}
$$

Where:

- $X$ = number of events in the interval
- $k$ = specific number of events
- $\lambda$ = average number of events in the interval
- $e$ = mathematical constant approximately equal to 2.71828
- $k!$ = factorial of k

### Where it is used
It is used for count data such as customer arrivals, phone calls, website requests, or equipment failures during a fixed interval.

### Real-world Example
Suppose a customer-support center receives an average of 5 calls per minute.

Then:

$$
\lambda=5
$$

Poisson distribution can estimate the probability of receiving a particular number of calls in one minute.

### AI/ML Example
Poisson distribution can be used when the target variable represents counts.

Examples include predicting the number of website visits, customer arrivals, claims, or events during a time interval.

### Python Visualization


```python
import numpy as np
import matplotlib.pyplot as plt

lam = 5
data = np.random.poisson(lam, 1000)

values, counts = np.unique(data, return_counts=True)

plt.bar(values, counts / len(data))
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

### Explanation
The code generates event counts using an average rate of 5 events per interval.

The visualization shows the probability pattern of different event counts. Poisson-based models are useful in ML when the target is a non-negative count.


# 4. Uniform Distribution


### Definition
Uniform distribution is a distribution where every value within a specified continuous interval has the same probability density.

### Formula

$$
f(x)=\frac{1}{b-a}, \quad a\le x\le b
$$

Where:

- $f(x)$ = probability density
- $x$ = individual value
- $a$ = minimum value or lower limit
- $b$ = maximum value or upper limit

### Where it is used
It is used when values within a range are considered equally likely.

It is also commonly used for random sampling and simulation.

### Real-world Example
Suppose a random number is selected between 0 and 10 from a continuous uniform distribution.

Every equal-length interval inside 0 to 10 has the same probability.

### AI/ML Example
Uniform distribution can be used to sample hyperparameter values during randomized experiments.

For example, learning-rate values can be sampled from a chosen interval.

### Python Visualization


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Uniform Distribution")
plt.show()
```

### Explanation
The code generates 1000 random values between 0 and 10.

The histogram is approximately flat because equal-sized intervals have approximately equal probability. Uniform sampling is useful when we want to explore values across a range without favouring one region.


# 5. Normal Distribution


### Definition
Normal distribution is a continuous, symmetric, bell-shaped distribution where values are concentrated around the mean.

It is also called Gaussian distribution.

### Formula

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- $x$ = individual observed value
- $\mu$ = population mean or average
- $\sigma$ = population standard deviation
- $\pi$ = mathematical constant approximately equal to 3.14159
- $e$ = mathematical constant approximately equal to 2.71828

### Where it is used
It is used to model many approximately symmetric continuous measurements and appears in statistical inference and modelling assumptions.

### Real-world Example
Human heights in a sufficiently homogeneous population may be approximately normally distributed.

Most people are near the average height, while extremely short or tall values are less common.

### AI/ML Example
Normal distribution is often used when examining model residuals and in probabilistic models that assume Gaussian noise.

It is also related to standardization through the mean and standard deviation.

### Python Visualization


```python
import numpy as np
import matplotlib.pyplot as plt

mu = 50
sigma = 10

data = np.random.normal(mu, sigma, 1000)

plt.hist(data, bins=30, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

### Explanation
The code generates 1000 values with population mean 50 and population standard deviation 10.

The resulting histogram forms an approximate bell shape. Understanding normal distributions is useful for statistical modelling, residual analysis, standardization, and several probabilistic ML methods.


# 6. Exponential Distribution


### Definition
Exponential distribution models the waiting time between events when the events follow a Poisson process with a constant average rate.

### Formula

$$
f(x)=\lambda e^{-\lambda x}, \quad x\ge0
$$

Where:

- $f(x)$ = probability density
- $x$ = waiting time
- $\lambda$ = event rate per unit of time
- $e$ = mathematical constant approximately equal to 2.71828

The mean waiting time is:

$$
E(X)=\frac{1}{\lambda}
$$

Where:

- $E(X)$ = expected or average waiting time
- $\lambda$ = average event rate

### Where it is used
It is used for waiting-time problems such as time between customer arrivals, machine failures, calls, or service requests when a constant-rate Poisson-process assumption is reasonable.

### Real-world Example
Suppose customers arrive at an average rate of 2 customers per minute.

Then:

$$
\lambda=2
$$

The expected waiting time is:

$$
E(X)=\frac{1}{2}=0.5
$$

So the average waiting time is 0.5 minutes under the model.

### AI/ML Example
Exponential distributions can be useful in probabilistic modelling of waiting times and time-to-event data under constant-hazard assumptions.

They may also appear in simulation and reliability-related ML applications.

### Python Visualization


```python
import numpy as np
import matplotlib.pyplot as plt

lam = 2
scale = 1 / lam

data = np.random.exponential(scale, 1000)

plt.hist(data, bins=30, density=True)
plt.xlabel("Waiting Time")
plt.ylabel("Density")
plt.title("Exponential Distribution")
plt.show()
```

### Explanation
The code generates 1000 waiting times using an event rate of 2 per unit time.

NumPy uses the scale parameter, which equals $1/\lambda$.

The distribution has many short waiting times and fewer long waiting times. This makes it useful for modelling certain time-to-event and waiting-time problems.
