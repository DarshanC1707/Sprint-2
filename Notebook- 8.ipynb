# Correlation & Covariance




# 1. Correlation

### Definition

Correlation measures the direction and strength of the relationship between two variables.

### Real-life Example

Study hours and exam marks may increase together.

### Formula / Numerical Example

$$r=\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i-\bar{x})^2\sum_{i=1}^{n}(y_i-\bar{y})^2}}$$

Where:
- $r$ = correlation coefficient
- $x_i$ = each individual X value
- $y_i$ = each individual Y value
- $\bar{x}$ = sample mean of X
- $\bar{y}$ = sample mean of Y
- $n$ = number of paired observations

For X = [1, 2, 3] and Y = [2, 4, 6], $r=1$, showing perfect positive linear correlation.

### Python Implementation


```python
import pandas as pd
x = pd.Series([1,2,3,4,5])
y = pd.Series([2,4,6,8,10])
print("Correlation:", x.corr(y))
```

### AI/ML Example

Correlation is useful in AI/ML exploratory data analysis for understanding relationships among numerical features. The code calculates the linear correlation between X and Y. Correlation does not prove causation.


# 2. Covariance

### Definition

Covariance measures whether two variables tend to move in the same or opposite directions.

### Real-life Example

Temperature and ice-cream sales may increase together, producing positive covariance.

### Formula / Numerical Example

$$\operatorname{Cov}(X,Y)=\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{n-1}$$

Where:
- $\operatorname{Cov}(X,Y)$ = sample covariance
- $x_i$, $y_i$ = each paired value
- $\bar{x}$, $\bar{y}$ = sample means
- $n$ = number of paired observations

For X = [1,2,3] and Y = [2,4,6], sample covariance is 2.

### Python Implementation


```python
import numpy as np
x=[1,2,3]
y=[2,4,6]
print("Covariance:", np.cov(x,y,ddof=1)[0,1])
```

### AI/ML Example

Covariance helps study how numerical features vary together and is important in methods such as PCA. Its magnitude depends on the units and scales of the variables.


# 3. Positive Correlation

### Definition

Positive correlation occurs when two variables tend to move in the same direction.

### Real-life Example

As study hours increase, exam marks may also increase.

### Formula / Numerical Example

$$0<r\le1$$

Where:
- $r$ = correlation coefficient
- values closer to +1 indicate stronger positive linear correlation

For X = [1,2,3,4] and Y = [10,20,30,40], $r=1$.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([1,2,3,4,5])
y=np.array([10,20,30,40,50])
print("Correlation:",np.corrcoef(x,y)[0,1])
plt.scatter(x,y)
plt.xlabel("X"); plt.ylabel("Y"); plt.title("Positive Correlation")
plt.show()
```

### AI/ML Example

A numerical feature can have positive correlation with a numerical target. The scatter plot rises from left to right because both variables increase together.


# 4. Negative Correlation

### Definition

Negative correlation occurs when two variables tend to move in opposite directions.

### Real-life Example

For a fixed journey distance, increasing speed generally reduces travel time.

### Formula / Numerical Example

$$-1\le r<0$$

Where:
- $r$ = correlation coefficient
- values closer to -1 indicate stronger negative linear correlation

For X = [1,2,3,4] and Y = [40,30,20,10], $r=-1$.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([1,2,3,4,5])
y=np.array([50,40,30,20,10])
print("Correlation:",np.corrcoef(x,y)[0,1])
plt.scatter(x,y)
plt.xlabel("X"); plt.ylabel("Y"); plt.title("Negative Correlation")
plt.show()
```

### AI/ML Example

A feature may have a negative relationship with a numerical target. The scatter plot falls from left to right because one variable increases while the other decreases.


# 5. No Correlation

### Definition

No correlation means there is little or no linear relationship between two variables.

### Real-life Example

Shoe size and exam marks would generally not be expected to have a meaningful linear relationship.

### Formula / Numerical Example

$$r\approx0$$

Where:
- $r$ = correlation coefficient
- a value near 0 indicates little or no linear relationship

A correlation near zero does not rule out a nonlinear relationship.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([1,2,3,4,5,6])
y=np.array([4,1,6,2,5,3])
print("Correlation:",np.corrcoef(x,y)[0,1])
plt.scatter(x,y)
plt.xlabel("X"); plt.ylabel("Y"); plt.title("No Clear Linear Correlation")
plt.show()
```

### AI/ML Example

In AI/ML, a feature can have weak linear correlation with a target while still having a nonlinear relationship. The scatter plot helps visualize the absence of a clear linear direction.


# 6. Pearson Correlation

### Definition

Pearson correlation measures the strength and direction of a linear relationship between two numerical variables.

### Real-life Example

It can measure the linear relationship between study hours and exam scores.

### Formula / Numerical Example

$$r=\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i-\bar{x})^2\sum_{i=1}^{n}(y_i-\bar{y})^2}}$$

Where:
- $r$ = Pearson correlation coefficient
- $x_i$, $y_i$ = each paired value
- $\bar{x}$, $\bar{y}$ = sample means
- $n$ = number of paired observations

$r=1$ indicates perfect positive linear correlation and $r=-1$ perfect negative linear correlation.

### Python Implementation


```python
from scipy.stats import pearsonr
x=[1,2,3,4,5]
y=[2,4,5,8,10]
r,p=pearsonr(x,y)
print("Pearson Correlation:",r)
print("P-value:",p)
```

### AI/ML Example

Pearson correlation is commonly used in exploratory data analysis for continuous numerical features. It focuses on linear relationships and can be strongly affected by outliers.


# 7. Spearman Correlation

### Definition

Spearman correlation measures the strength and direction of a monotonic relationship using ranks.

### Real-life Example

It can compare the ranking of students by study effort with their ranking by exam performance.

### Formula / Numerical Example

When there are no tied ranks:

$$\rho=1-\frac{6\sum d_i^2}{n(n^2-1)}$$

Where:
- $\rho$ = Spearman rank correlation
- $d_i$ = difference between the two ranks for observation i
- $n$ = number of paired observations

If all paired ranks are identical, every $d_i=0$ and $\rho=1$.

### Python Implementation


```python
from scipy.stats import spearmanr
x=[1,2,3,4,5]
y=[10,20,40,80,160]
rho,p=spearmanr(x,y)
print("Spearman Correlation:",rho)
print("P-value:",p)
```

### AI/ML Example

Spearman correlation is useful for ordinal data or monotonic relationships that are not necessarily linear. It works with ranks rather than the exact spacing between values.


# 8. Correlation Matrix

### Definition

A correlation matrix is a table containing pairwise correlation coefficients between multiple numerical variables.

### Real-life Example

A house dataset may compare Area, Bedrooms, Age, and Price in one matrix.

### Formula / Numerical Example

$$R=\begin{bmatrix}1&r_{12}&r_{13}\\r_{21}&1&r_{23}\\r_{31}&r_{32}&1\end{bmatrix}$$

Where:
- $R$ = correlation matrix
- $r_{ij}$ = correlation between variable i and variable j
- diagonal values are 1 because each variable is perfectly correlated with itself

### Python Implementation


```python
import pandas as pd
df=pd.DataFrame({
"Area":[1000,1200,1500,1800,2000],
"Bedrooms":[2,2,3,3,4],
"Age":[20,18,15,10,5],
"Price":[40,45,60,75,90]
})
corr=df.corr(numeric_only=True)
print(corr)
```

### AI/ML Example

A correlation matrix helps inspect many numerical feature relationships at once. It can highlight strongly related features and possible multicollinearity, but correlation alone should not decide feature removal.


# 9. Heatmap

### Definition

A heatmap visually represents numerical matrix values using colors or shades.

### Real-life Example

A correlation heatmap makes a large correlation matrix easier to inspect.

### Formula / Numerical Example

A heatmap has no separate statistical formula. For a correlation heatmap it visualizes:

$$-1\le r\le1$$

Where:
- $r$ = correlation coefficient
- $-1$ = perfect negative linear correlation
- $0$ = no linear correlation
- $+1$ = perfect positive linear correlation

### Python Implementation


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "Sleep": [8, 7, 7, 6, 5]
}

df = pd.DataFrame(data)

corr = df.corr()

sns.heatmap(corr, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

This code creates a DataFrame with Hours, Marks, and Sleep, then calculates the correlation between these variables. The heatmap visually displays the correlation values, making it easy to understand which variables have positive, negative, or weak relationships.

### AI/ML Example

Correlation heatmaps are useful during exploratory data analysis. The code calculates a correlation matrix and visualizes all pairwise relationships so strong positive and negative correlations are easier to inspect.
