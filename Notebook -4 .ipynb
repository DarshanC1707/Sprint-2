# Data Distribution


# 1. Normal Distribution


### Definition
Normal distribution is a continuous probability distribution where most values are concentrated around the mean and fewer values occur as we move away from the mean.

It has a bell-shaped curve.

### Real-life Example
Suppose the heights of a large group of people are measured.

Most people may have heights close to the average height, while very short and very tall people are less common.

### Formula

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- $\mu$ = population mean or average
- $\sigma$ = population standard deviation
- $x$ = individual observed value
- $\pi$ = mathematical constant approximately equal to 3.14159
- $e$ = mathematical constant approximately equal to 2.71828

### Numerical Example
Suppose:

Mean = 50  
Standard deviation = 10  
Value = 60

First calculate the z-score:

$$
z = \frac{x-\mu}{\sigma}
$$

Where:

- $z$ = number of standard deviations the value is from the mean
- $x$ = individual observed value
- $\mu$ = population mean or average
- $\sigma$ = population standard deviation

$$
z = \frac{60-50}{10}=1
$$

The value 60 is one standard deviation above the mean.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30, density=True)
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

### AI/ML Example
Normal distribution is commonly checked when analysing numerical features.


```python
import numpy as np
import matplotlib.pyplot as plt

errors = np.random.normal(0, 1, 1000)

plt.hist(errors, bins=30)
plt.xlabel("Error")
plt.ylabel("Frequency")
plt.title("Distribution of Model Errors")
plt.show()
```

### Explanation
This code generates example model errors centered around zero and displays their distribution.

In some statistical and machine learning models, assumptions are made about errors being approximately normally distributed. Examining the error distribution can therefore help with model diagnostics and understanding model behaviour.


# 2. Uniform Distribution


### Definition
Uniform distribution is a probability distribution where every value within a specified interval has the same probability density.

### Real-life Example
Consider a fair six-sided dice.

Each number from 1 to 6 has the same probability of appearing.

### Formula

$$
f(x)=\frac{1}{b-a}
$$

for:

$$
a \le x \le b
$$

Where:

- $f(x)$ = probability density
- $a$ = minimum value or lower limit
- $b$ = maximum value or upper limit
- $x$ = individual value within the interval

### Numerical Example
Suppose:

$$
a=0
$$

$$
b=10
$$

Then:

$$
f(x)=\frac{1}{10-0}
$$

$$
f(x)=0.1
$$

Every value between 0 and 10 has the same probability density.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20, density=True)
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Uniform Distribution")
plt.show()
```

### AI/ML Example
Uniform distribution can be used when generating random values during machine learning experiments.


```python
import numpy as np

learning_rates = np.random.uniform(
    low=0.001,
    high=0.1,
    size=5
)

print(learning_rates)
```

### Explanation
The code randomly generates five learning-rate values between 0.001 and 0.1.

Uniform sampling gives values across the selected interval an equal chance of being sampled. This can be useful in randomized experiments and some forms of hyperparameter search.


# 3. Gaussian Distribution


### Definition
Gaussian distribution is another name for the normal distribution.

It is a symmetric, bell-shaped probability distribution where values are concentrated around the mean.

### Real-life Example
Measurement errors from repeated experiments often approximately follow a Gaussian distribution when many small independent effects contribute to the error.

Most errors are close to zero, while very large positive or negative errors are less common.

### Formula

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- $\mu$ = population mean or average
- $\sigma$ = population standard deviation
- $x$ = individual observed value
- $\pi$ = mathematical constant approximately equal to 3.14159
- $e$ = mathematical constant approximately equal to 2.71828

### Numerical Example
Suppose:

$$
\mu=100
$$

$$
\sigma=15
$$

$$
x=115
$$

Calculate the z-score:

$$
z=\frac{x-\mu}{\sigma}
$$

Where:

- $z$ = number of standard deviations from the mean
- $x$ = individual observed value
- $\mu$ = population mean or average
- $\sigma$ = population standard deviation

$$
z=\frac{115-100}{15}=1
$$

Therefore, 115 is one standard deviation above the mean.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=100, scale=15, size=1000)

plt.hist(data, bins=30, density=True)
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Gaussian Distribution")
plt.show()
```

### AI/ML Example
Gaussian noise can be added to data during experiments or data augmentation.


```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

noise = np.random.normal(
    loc=0,
    scale=1,
    size=len(data)
)

noisy_data = data + noise

print(noisy_data)
```

### Explanation
The code generates small random values from a Gaussian distribution and adds them to the original data.

Adding controlled noise can sometimes be used for data augmentation or robustness experiments, helping evaluate how a model behaves when input data contains small variations.


# 4. Left Skew


### Definition
A left-skewed distribution has a longer tail on the left side of the distribution.

It is also called negatively skewed distribution.

A common relationship is:

$$
\text{Mean} < \text{Median} < \text{Mode}
$$

This relationship is a general pattern, not a rule that must hold for every negatively skewed dataset.

### Real-life Example
Consider marks from an easy examination.

Most students may score high marks, while only a few students receive very low marks.

The few low scores create a tail toward the left.

### Formula

$$
\text{Skewness}
=
E\left[
\left(
\frac{X-\mu}{\sigma}
\right)^3
\right]
$$

Where:

- $E$ = expected value or average over the population
- $X$ = random variable or observed values
- $\mu$ = population mean or average
- $\sigma$ = population standard deviation

For left skew:

$$
\text{Skewness} < 0
$$

### Numerical Example

$$
\text{Skewness}=-1.2
$$

Since the value is negative, the distribution is negatively skewed.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

data = -np.random.exponential(scale=2, size=1000)

print("Skewness:", skew(data))

plt.hist(data, bins=30)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Left-Skewed Distribution")
plt.show()
```

### AI/ML Example
We can check the skewness of a numerical feature before model training.


```python
import pandas as pd

df = pd.DataFrame({
    "Score": [20, 70, 75, 80, 82, 85, 88, 90]
})

skewness = df["Score"].skew()

print("Skewness:", skewness)
```

### Explanation
The code calculates the skewness of the Score feature.

A negative skewness value indicates that the distribution has more of its tail toward lower values.

Understanding skewness helps during exploratory data analysis because highly skewed features may sometimes benefit from transformation depending on the model and the problem.


# 5. Right Skew


### Definition
A right-skewed distribution has a longer tail on the right side.

It is also called positively skewed distribution.

A common relationship is:

$$
\text{Mode} < \text{Median} < \text{Mean}
$$

This is a common pattern rather than a universal rule.

### Real-life Example
Income is often right-skewed.

Many people may have low or moderate incomes, while a small number of people have extremely high incomes.

These high values create the long right tail.

### Formula

For right skew:

$$
\text{Skewness}>0
$$

Where:

- Skewness = measure of the asymmetry of a distribution
- Positive skewness = longer tail toward higher values

### Numerical Example

$$
\text{Skewness}=1.5
$$

Because the skewness is positive, the distribution is right-skewed.

### Python Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

data = np.random.exponential(scale=2, size=1000)

print("Skewness:", skew(data))

plt.hist(data, bins=30)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Right-Skewed Distribution")
plt.show()
```

### AI/ML Example
Suppose a dataset contains customer income.


```python
import pandas as pd

df = pd.DataFrame({
    "Income": [25000, 28000, 30000, 32000, 35000, 40000, 50000, 200000]
})

print("Skewness:", df["Income"].skew())
```

### Explanation
The code calculates the skewness of the Income feature.

The very high income of 200000 stretches the distribution toward the right, producing positive skewness.

Identifying highly skewed numerical features can help us decide whether transformations, robust statistics, or other preprocessing techniques may be appropriate.


# 6. Symmetric Distribution


### Definition
A symmetric distribution has approximately the same shape on both sides of its center.

For a perfectly symmetric unimodal distribution such as a normal distribution:

$$
\text{Mean}=\text{Median}=\text{Mode}
$$

### Real-life Example
Suppose exam scores are distributed around 70.

Approximately equal numbers of students score above and below 70, producing a roughly symmetric shape.

### Formula

For a perfectly symmetric distribution:

$$
\text{Skewness}=0
$$

Where:

- Skewness = measure of distribution asymmetry
- 0 = no left or right skew in a perfectly symmetric distribution

### Numerical Example
Consider:

10, 20, 30, 40, 50

The center is 30.

$$
30-10=20
$$

$$
50-30=20
$$

$$
30-20=10
$$

$$
40-30=10
$$

The values are equally spread around the center.

### Python Implementation


```python
import numpy as np
from scipy.stats import skew

data = [10, 20, 30, 40, 50]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Skewness:", skew(data))
```

### AI/ML Example
We can check whether a numerical feature is approximately symmetric before preprocessing.


```python
import numpy as np
from scipy.stats import skew

feature = np.random.normal(loc=50, scale=10, size=1000)

print("Skewness:", skew(feature))
```

### Explanation
The code generates approximately normally distributed values and calculates their skewness.

For a large sample from a symmetric distribution, the calculated skewness should usually be reasonably close to zero.

Understanding symmetry helps us describe feature distributions and choose suitable statistical or preprocessing methods.


# 7. Kurtosis


### Definition
Kurtosis measures how much probability a distribution places in its tails relative to a normal distribution.

In practice, higher kurtosis is associated with heavier tails and a greater tendency to produce extreme observations.

### Real-life Example
Consider daily financial returns.

Some datasets may usually have small changes but occasionally contain very large gains or losses.

Such extreme observations can produce high kurtosis.

### Formula

$$
\text{Kurtosis}
=
E\left[
\left(
\frac{X-\mu}{\sigma}
\right)^4
\right]
$$

Where:

- $E$ = expected value or population average
- $X$ = random variable or observed values
- $\mu$ = population mean or average
- $\sigma$ = population standard deviation

Excess kurtosis is:

$$
\text{Excess Kurtosis}
=
\text{Kurtosis}-3
$$

For a normal distribution:

$$
\text{Kurtosis}=3
$$

$$
\text{Excess Kurtosis}=0
$$

### Numerical Example
Suppose:

$$
\text{Kurtosis}=5
$$

Then:

$$
\text{Excess Kurtosis}=5-3=2
$$

The positive excess kurtosis indicates heavier tails than a normal distribution.

### Python Implementation


```python
import numpy as np
from scipy.stats import kurtosis

data = np.random.normal(loc=0, scale=1, size=1000)

value = kurtosis(data)

print("Excess Kurtosis:", value)
```

### AI/ML Example
Kurtosis can help examine whether a feature contains heavy tails or extreme observations.


```python
import pandas as pd

df = pd.DataFrame({
    "Transaction": [100, 110, 120, 125, 130, 135, 140, 1000]
})

print("Kurtosis:", df["Transaction"].kurt())
```

### Explanation
The code calculates the sample excess kurtosis of the Transaction feature.

The unusually large transaction can increase tail heaviness and affect the kurtosis value.

In data analysis, kurtosis can help identify features that may contain heavy tails or extreme observations that deserve further investigation.


# 8. Empirical Rule (68–95–99.7)


### Definition
The empirical rule describes how data is distributed around the mean when the data approximately follows a normal distribution.

Approximately:

68% of values are within 1 standard deviation.

95% of values are within 2 standard deviations.

99.7% of values are within 3 standard deviations.

### Real-life Example
Suppose exam marks approximately follow a normal distribution with:

Mean = 70  
Standard deviation = 10

Most students' marks will fall within predictable ranges around 70.

### Formula

$$
\mu \pm 1\sigma \approx 68\%
$$

$$
\mu \pm 2\sigma \approx 95\%
$$

$$
\mu \pm 3\sigma \approx 99.7\%
$$

Where:

- $\mu$ = population mean or average
- $\sigma$ = population standard deviation
- $\pm$ = values below and above the mean

### Numerical Example
Given:

$$
\mu=70
$$

$$
\sigma=10
$$

68% range:

$$
70 \pm 10 = 60 \text{ to } 80
$$

95% range:

$$
70 \pm 20 = 50 \text{ to } 90
$$

99.7% range:

$$
70 \pm 30 = 40 \text{ to } 100
$$

### Python Implementation


```python
mean = 70
std = 10

range_68 = (mean - std, mean + std)
range_95 = (mean - 2 * std, mean + 2 * std)
range_997 = (mean - 3 * std, mean + 3 * std)

print("68%:", range_68)
print("95%:", range_95)
print("99.7%:", range_997)
```

### AI/ML Example
The empirical rule can provide a simple way to inspect unusually distant values when a feature is approximately normally distributed.


```python
import numpy as np

data = np.array([48, 49, 50, 51, 52, 49, 50, 100])

mean = np.mean(data)
std = np.std(data)

lower = mean - 3 * std
upper = mean + 3 * std

unusual_values = data[(data < lower) | (data > upper)]

print("Mean:", mean)
print("Standard Deviation:", std)
print("Unusual Values:", unusual_values)
```

### Explanation
The code calculates the mean and standard deviation and then creates boundaries three standard deviations away from the mean.

For approximately normal data, observations beyond three standard deviations are uncommon.

This idea can help flag unusual observations, but the empirical rule should mainly be used when the distribution is reasonably close to normal.
