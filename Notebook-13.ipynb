# Statistical Data Visualization


# 1. Histogram

### Purpose

Shows the frequency distribution of numerical data using bins.

### When to use

Use it to study distribution, spread and skewness of numerical data.

### Advantages

Simple and useful for understanding distribution shape.

### Limitations

The appearance depends on bin size and exact values are not shown.

### Python Example


```python
import matplotlib.pyplot as plt
marks=[45,50,52,55,60,62,65,68,70,72,75,80]
plt.hist(marks,bins=5,edgecolor="black")
plt.xlabel("Marks"); plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```

### Interpretation

The histogram groups marks into intervals and shows how many values fall in each interval. It helps us understand the overall distribution of marks.


# 2. Bar Chart

### Purpose

Compares values across different categories using bars.

### When to use

Use it for categorical data such as products, departments or regions.

### Advantages

Easy to compare categories and differences in magnitude.

### Limitations

It can become crowded with many categories and is not suitable for showing a continuous distribution.

### Python Example


```python
import matplotlib.pyplot as plt
products=["A","B","C","D"]
sales=[120,180,150,210]
plt.bar(products,sales)
plt.xlabel("Product"); plt.ylabel("Sales")
plt.title("Product Sales")
plt.show()
```

### Interpretation

The chart compares sales across products. Taller bars represent products with higher sales.


# 3. Line Chart

### Purpose

Shows how a numerical value changes across an ordered sequence, commonly time.

### When to use

Use it to visualize trends and changes over time.

### Advantages

Clearly shows increases, decreases and trends.

### Limitations

It is not ideal for unrelated categories, and too many lines can make it difficult to read.

### Python Example


```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[100,120,115,140,160]
plt.plot(months,sales,marker="o")
plt.xlabel("Month"); plt.ylabel("Sales")
plt.title("Monthly Sales Trend")
plt.show()
```

### Interpretation

The line shows how sales change month by month, making the overall trend easy to observe.


# 4. Scatter Plot

### Purpose

Displays the relationship between two numerical variables using points.

### When to use

Use it to examine relationships, clusters, patterns and possible outliers.

### Advantages

Useful for seeing positive, negative, nonlinear or weak relationships.

### Limitations

Overlapping points can make large datasets difficult to read, and correlation does not prove causation.

### Python Example


```python
import matplotlib.pyplot as plt
hours=[1,2,3,4,5,6]
marks=[40,50,55,65,75,85]
plt.scatter(hours,marks)
plt.xlabel("Study Hours"); plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

### Interpretation

The points show that marks generally increase as study hours increase, indicating a positive relationship in this sample.


# 5. Pie Chart

### Purpose

Shows how categories contribute to a whole using slices of a circle.

### When to use

Use it for a small number of categories representing parts of a total.

### Advantages

Provides a quick visual view of proportions.

### Limitations

Similar slices are difficult to compare and too many categories make it unclear.

### Python Example


```python
import matplotlib.pyplot as plt
labels=["Product A","Product B","Product C"]
sales=[40,35,25]
plt.pie(sales,labels=labels,autopct="%1.1f%%")
plt.title("Sales Share")
plt.show()
```

### Interpretation

The pie chart shows each product's percentage share of total sales. Larger slices represent larger proportions.


# 6. Density Plot

### Purpose

Shows a smooth estimate of the distribution of numerical data.

### When to use

Use it to understand the shape and concentration of continuous data.

### Advantages

Provides a smooth view of distribution shape and helps compare distributions.

### Limitations

The result depends on smoothing bandwidth, and the vertical axis represents density rather than raw frequency.

### Python Example


```python
import seaborn as sns
import matplotlib.pyplot as plt
marks=[45,50,52,55,60,62,65,68,70,72,75,80]
sns.kdeplot(marks,fill=True)
plt.xlabel("Marks")
plt.title("Density Plot of Marks")
plt.show()
```

### Interpretation

The density curve shows where marks are more concentrated. Higher parts of the curve represent ranges with greater estimated density.


# 7. Box Plot

### Purpose

Summarizes a numerical distribution using median, quartiles, spread and potential outliers.

### When to use

Use it to inspect spread, compare distributions and identify possible outliers.

### Advantages

Compactly shows important distribution information.

### Limitations

It does not show every observation or the complete shape of the distribution.

### Python Example


```python
import matplotlib.pyplot as plt
data=[10,12,13,14,15,16,18,30]
plt.boxplot(data)
plt.ylabel("Values")
plt.title("Box Plot")
plt.show()
```

### Interpretation

The box represents the middle portion of the data, the line inside shows the median, and values beyond the whiskers may appear as potential outliers.


# 8. Violin Plot

### Purpose

Combines a distribution density shape with summary information to show numerical data.

### When to use

Use it to compare the shape and spread of numerical distributions across groups.

### Advantages

Shows more distribution-shape information than a standard box plot.

### Limitations

It can be less intuitive and may be misleading with very small samples.

### Python Example


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
df=pd.DataFrame({
    "Group":["A"]*6+["B"]*6,
    "Score":[45,50,55,60,62,65,55,60,65,70,75,80]
})
sns.violinplot(data=df,x="Group",y="Score")
plt.title("Score Distribution by Group")
plt.show()
```

### Interpretation

The width of each violin shows where values are more concentrated, helping us compare the shapes and spreads of the groups.


# 9. Pair Plot

### Purpose

Displays pairwise relationships among several numerical variables and their individual distributions.

### When to use

Use it during exploratory data analysis to examine several feature relationships at once.

### Advantages

Provides multiple relationships and distributions in one visualization.

### Limitations

It becomes crowded and slower when there are many features or observations.

### Python Example


```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
df=pd.DataFrame({
    "Hours":[1,2,3,4,5],
    "Marks":[40,50,60,70,80],
    "Sleep":[8,7,7,6,5]
})
sns.pairplot(df)
plt.show()
```

### Interpretation

The pair plot shows each variable's distribution and pairwise relationships among Hours, Marks and Sleep, helping us identify useful patterns.


# 10. Heatmap

### Purpose

Uses color intensity to represent numerical values in a matrix or table.

### When to use

Use it for correlation matrices, confusion matrices and other two-dimensional numerical tables.

### Advantages

Makes matrices easier to scan and reveals strong or weak patterns.

### Limitations

Exact values can be harder to compare without annotations, and interpretation depends on the scale.

### Python Example


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data={
    "Hours":[1,2,3,4,5],
    "Marks":[40,50,60,70,80],
    "Sleep":[8,7,7,6,5]
}
df=pd.DataFrame(data)
corr=df.corr()
sns.heatmap(corr,annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

### Interpretation

The heatmap visually shows the correlation between different variables, making it easy to identify strong positive, negative, or weak relationships.
