## Topic 4: Bayes' Theorem

### Task: Calculate the posterior probability using Bayes' Theorem.


```python
# Prior Probability
P_A = 0.25

# Likelihood
P_B_given_A = 0.70

# Evidence
P_B = 0.40

# Posterior Probability
P_A_given_B = (P_B_given_A * P_A) / P_B

print("Posterior Probability:", P_A_given_B)
```

## Topic 6: Correlation and Covariance

### Task: Calculate Correlation and Covariance between two variables.


```python
import numpy as np

x = [6, 12, 18, 24, 30]
y = [9, 17, 28, 36, 47]

correlation = np.corrcoef(x, y)[0, 1]
covariance = np.cov(x, y)[0, 1]

print("Correlation:", correlation)
print("Covariance:", covariance)
```

## Topic 8: Interquartile Range (IQR)

### Task: Calculate Q1, Q3, and IQR for the given dataset.


```python
import numpy as np

data = [6, 9, 11, 14, 17, 21, 24, 28, 33, 37]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

## 12. Detect Outliers Using the IQR Method

### Task: Detect outliers in the given dataset using the IQR Method.


```python
import numpy as np

data = [14, 16, 19, 21, 23, 26, 29, 85]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_limit or x > upper_limit]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:", outliers)
```
