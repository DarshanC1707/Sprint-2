#  Statistical Tests


# 1. T-Test

### Definition

A T-Test compares means when the population standard deviation is unknown.

### Assumptions

- Independent observations for independent-group tests.
- Numerical outcome.
- For small samples, data or paired differences should be approximately normal.
- Standard independent T-Test assumes equal variances; Welch's T-Test relaxes this.

### When to use

Use it for one sample versus a known value, two independent group means, or paired before-and-after means.

### When not to use

Do not use an independent T-Test for categorical outcomes, more than two groups, or paired observations.

### Business Example

A company compares average sales between two stores.

### Python Example


```python
from scipy.stats import ttest_ind
a=[50,52,48,55,53]
b=[60,62,58,61,63]
t,p=ttest_ind(a,b,equal_var=False)
print("T-statistic:",t)
print("p-value:",p)
```

### Interpretation

If p < 0.05, reject the null hypothesis of equal means at the 5% significance level; otherwise, fail to reject it.


# 2. Z-Test

### Definition

A Z-Test tests a population mean when the population standard deviation is known and normal or large-sample conditions are appropriate.

### Assumptions

- Independent observations.
- Numerical outcome.
- Population standard deviation is known for the classical mean Z-Test.
- Population is normal or sample size is sufficiently large for a normal approximation.

### When to use

Use it to test a population mean when the population standard deviation is known.

### When not to use

Do not use the classical mean Z-Test when population standard deviation is unknown and the sample is small; a T-Test is generally more suitable.

### Business Example

A manufacturer tests whether average package weight differs from 500 grams.

### Python Example


```python
import numpy as np
from scipy.stats import norm
sample_mean=505
mu=500
sigma=15
n=100
z=(sample_mean-mu)/(sigma/np.sqrt(n))
p=2*(1-norm.cdf(abs(z)))
print("Z-statistic:",z)
print("p-value:",p)
```

### Interpretation

For a two-tailed test, p < 0.05 provides evidence against the claimed population mean at the 5% significance level.


# 3. Chi-Square Test

### Definition

A Chi-Square Test works with categorical counts to test association between categorical variables or agreement between observed and expected frequencies.

### Assumptions

- Data are frequency counts.
- Observations are independent.
- Categories are mutually exclusive.
- Expected frequencies should generally be sufficiently large for the Chi-Square approximation.

### When to use

Use it for relationships between categorical variables or observed versus expected category frequencies.

### When not to use

Do not use it to compare numerical means. With very small expected counts, an exact or other suitable test may be better.

### Business Example

A business checks whether product preference is associated with customer region.

### Python Example


```python
import pandas as pd
from scipy.stats import chi2_contingency
table=pd.DataFrame({"A":[30,20],"B":[15,35]},index=["Region 1","Region 2"])
chi2,p,dof,expected=chi2_contingency(table)
print("Chi-Square:",chi2)
print("p-value:",p)
print("Expected:\n",expected)
```

### Interpretation

If p < 0.05, reject the null hypothesis of independence and conclude that the data provide evidence of an association.


# 4. ANOVA

### Definition

ANOVA means Analysis of Variance. One-way ANOVA tests whether the means of three or more independent groups are all equal.

### Assumptions

- Independent observations.
- Numerical dependent variable.
- Residuals within groups are approximately normal, especially for small samples.
- Classical one-way ANOVA assumes similar variances across groups.

### When to use

Use it to compare one numerical outcome across three or more independent groups.

### When not to use

Do not use ordinary one-way ANOVA for categorical outcomes or repeated measurements on the same subjects.

### Business Example

A company compares average sales across three marketing strategies.

### Python Example


```python
from scipy.stats import f_oneway
a=[50,52,48,51,49]
b=[60,62,58,61,59]
c=[70,72,68,71,69]
f,p=f_oneway(a,b,c)
print("F-statistic:",f)
print("p-value:",p)
```

### Interpretation

If p < 0.05, at least one group mean differs. A post-hoc analysis is needed to identify which groups differ.


# 5. Mann-Whitney U Test

### Definition

The Mann-Whitney U Test is a rank-based non-parametric test for comparing two independent groups.

### Assumptions

- Two independent groups.
- Ordinal or continuous observations.
- Independent observations within groups.
- Similar distribution shapes are needed if interpreting the result specifically as a location or median difference.

### When to use

Use it when comparing two independent groups with ordinal data or when a rank-based alternative to a T-Test is appropriate.

### When not to use

Do not use it for paired data or more than two independent groups.

### Business Example

A company compares satisfaction ratings from two independent branches.

### Python Example


```python
from scipy.stats import mannwhitneyu
a=[2,3,3,4,4]
b=[4,4,5,5,5]
u,p=mannwhitneyu(a,b,alternative="two-sided")
print("U-statistic:",u)
print("p-value:",p)
```

### Interpretation

If p < 0.05, reject the null hypothesis under the test's formulation. Interpret the result together with the shapes of the group distributions.


# 6. Shapiro-Wilk Test

### Definition

The Shapiro-Wilk Test checks whether sample data are consistent with a normal distribution.

### Assumptions

- Numerical variable.
- Independent observations.
- Sample represents the population or process of interest.
- For large samples, even small departures from normality may be detected.

### When to use

Use it to assess normality of numerical data or model residuals before methods that rely on normality assumptions.

### When not to use

Do not use it as the only evidence for normality; also inspect plots and practical importance.

### Business Example

A company checks whether delivery-time data are approximately normal before applying a parametric test.

### Python Example


```python
from scipy.stats import shapiro
x=[12.1,11.8,12.5,12.0,11.9,12.3,12.2,11.7]
stat,p=shapiro(x)
print("Statistic:",stat)
print("p-value:",p)
```

### Interpretation

If p < 0.05, there is evidence against normality. If p >= 0.05, fail to reject normality; this does not prove perfect normality.
