# Sampling


# 1. Sampling

### Definition

Sampling is the process of selecting a smaller group from a larger population to study and make conclusions about the population.

### Real-life Example

Suppose a college has 5,000 students. Instead of surveying all students, we may select 500 students and use their responses to understand the larger population.

### Formula / Numerical Example

$$
\hat{p}=\frac{x}{n}
$$

Where:

- $\hat{p}$ = sample proportion
- $x$ = number of sampled observations with the characteristic
- $n$ = sample size

If 60 out of 100 sampled customers are satisfied:

$$
\hat{p}=\frac{60}{100}=0.60
$$

### Python Implementation

```python
import pandas as pd

population = pd.DataFrame({"Customer_ID": range(1, 1001)})
sample = population.sample(n=100, random_state=42)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print(sample.head())
```

### AI/ML Example

In AI/ML, datasets are usually samples from a broader real-world population or data-generating process.

The code selects 100 observations from 1,000. Representative sampling is important because the quality of the sample affects how well a model generalizes to unseen data.

# 2. Random Sampling

### Definition

Random sampling is a method where each member of the population has an equal chance of being selected.

### Real-life Example

A company with 1,000 employees can randomly select 100 employee IDs for a feedback survey.

### Formula / Numerical Example

For simple random sampling without replacement:

$$
P(\text{selected})=\frac{n}{N}
$$

Where:

- $P(\text{selected})$ = probability that a particular member is included
- $n$ = sample size
- $N$ = population size

If $N=1000$ and $n=100$:

$$
P(\text{selected})=\frac{100}{1000}=0.10
$$

### Python Implementation

```python
import random

population = list(range(1, 101))
sample = random.sample(population, 10)

print("Random Sample:", sample)
```

### AI/ML Example

Random sampling can be used to create subsets of data without deliberately selecting particular observations.

The code randomly selects 10 unique values from 100. Random selection helps reduce systematic selection bias, although one random sample is not guaranteed to perfectly represent every population characteristic.

# 3. Stratified Sampling

### Definition

Stratified sampling divides the population into meaningful subgroups called strata and then samples from each subgroup.

### Real-life Example

A college can divide students by department and take a proportional sample from Computer Science, Electronics, and Mechanical departments.

### Formula / Numerical Example

For proportional stratified sampling:

$$
n_h=\frac{N_h}{N}\times n
$$

Where:

- $n_h$ = sample size from stratum h
- $N_h$ = population size of stratum h
- $N$ = total population size
- $n$ = total required sample size
- $h$ = a particular subgroup

If $N=1000$, $N_h=500$, and $n=100$:

$$
n_h=\frac{500}{1000}\times100=50
$$

### Python Implementation

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["CS"] * 50 + ["Electronics"] * 30 + ["Mechanical"] * 20,
    "Student_ID": range(1, 101)
})

sample = df.groupby("Department", group_keys=False).sample(
    frac=0.20, random_state=42
)

print(sample["Department"].value_counts())
```

### AI/ML Example

In classification tasks, stratified splitting is often used to preserve approximately the same class proportions in training and testing sets.

The code samples 20% from every department. Stratification is especially useful when important groups or target classes need representation in each sample.

# 4. Cluster Sampling

### Definition

Cluster sampling divides a population into natural groups called clusters and randomly selects entire clusters for study.

### Real-life Example

A researcher studying 100 schools may randomly select 10 schools and collect data from students in those selected schools.

### Formula / Numerical Example

If $C$ clusters exist and $c$ clusters are selected:

$$
P(\text{cluster selected})=\frac{c}{C}
$$

Where:

- $C$ = total number of clusters
- $c$ = number of selected clusters

If $C=100$ and $c=10$:

$$
P(\text{cluster selected})=\frac{10}{100}=0.10
$$

### Python Implementation

```python
import pandas as pd
import random

df = pd.DataFrame({
    "School": ["A"] * 5 + ["B"] * 5 + ["C"] * 5 + ["D"] * 5,
    "Student_ID": range(1, 21)
})

clusters = df["School"].unique().tolist()
selected = random.sample(clusters, 2)
sample = df[df["School"].isin(selected)]

print("Selected Clusters:", selected)
print(sample)
```

### AI/ML Example

Cluster sampling can be useful when AI/ML data is naturally grouped by locations, stores, hospitals, schools, or devices.

The code selects two schools and includes observations from those clusters. It can make data collection easier, although observations within one cluster may be similar to each other.

# 5. Systematic Sampling

### Definition

Systematic sampling selects observations at a fixed interval after choosing a random starting point.

### Real-life Example

A factory inspector may select a random starting product and then inspect every 10th product.

### Formula / Numerical Example

The sampling interval is:

$$
k=\frac{N}{n}
$$

Where:

- $k$ = sampling interval
- $N$ = population size
- $n$ = required sample size

If $N=1000$ and $n=100$:

$$
k=\frac{1000}{100}=10
$$

### Python Implementation

```python
import random

population = list(range(1, 101))
sample_size = 10
k = len(population) // sample_size

start = random.randint(0, k - 1)
sample = population[start::k][:sample_size]

print("Interval:", k)
print("Systematic Sample:", sample)
```

### AI/ML Example

Systematic sampling can reduce a large ordered dataset by selecting observations at regular intervals.

The code selects every kth observation after a random starting point. It should be used carefully when the data order contains repeating patterns that may align with the sampling interval.

# 6. Sampling Bias

### Definition

Sampling bias occurs when the sampling process systematically favors some members of the population over others.

### Real-life Example

An internet-usage survey conducted only through an online form may underrepresent people who rarely use the internet.

### Formula / Numerical Example

A general estimator-bias formula is:

$$
\text{Bias}(\hat{\theta})=E(\hat{\theta})-\theta
$$

Where:

- $\hat{\theta}$ = estimator calculated from a sample
- $E(\hat{\theta})$ = expected value of the estimator
- $\theta$ = true population parameter

If $\theta=50$ and $E(\hat{\theta})=55$:

$$
\text{Bias}=55-50=5
$$

### Python Implementation

```python
population_value = 50
expected_sample_estimate = 55

bias = expected_sample_estimate - population_value

print("Bias:", bias)
```

### AI/ML Example

In AI/ML, an unrepresentative training sample can cause a model to learn patterns that work better for well-represented cases than underrepresented ones.

The code demonstrates statistical estimator bias. In ML, sampling bias is mainly addressed through careful data collection, dataset auditing, and appropriate evaluation.

# 7. Sampling Error

### Definition

Sampling error is the natural difference between a statistic calculated from a sample and the true population parameter because only part of the population was observed.

### Real-life Example

If the population average height is 170 cm but a random sample has an average of 168 cm, the difference is sampling error.

### Formula / Numerical Example

For a sample mean:

$$
\text{Sampling Error}=\bar{x}-\mu
$$

Where:

- $\bar{x}$ = sample mean or average
- $\mu$ = population mean or average

The standard error of the sample mean is:

$$
SE=\frac{\sigma}{\sqrt{n}}
$$

Where:

- $SE$ = standard error
- $\sigma$ = population standard deviation
- $n$ = sample size

If $\bar{x}=168$ and $\mu=170$:

$$
\text{Sampling Error}=168-170=-2
$$

### Python Implementation

```python
population_mean = 170
sample_mean = 168

sampling_error = sample_mean - population_mean

print("Sampling Error:", sampling_error)
```

### AI/ML Example

Model evaluation metrics can vary depending on which observations appear in a test sample.

The code calculates the difference between a sample mean and population mean. In AI/ML, repeated sampling or cross-validation helps us understand how evaluation results vary across different samples.

# 8. Sample Size

### Definition

Sample size is the number of observations included in a sample. It is usually represented by n.

### Real-life Example

A customer survey with only 20 responses may give an unstable estimate. A properly selected larger sample can usually provide a more precise estimate.

### Formula / Numerical Example

For estimating a population proportion under simple random sampling:

$$
n=\frac{Z^2p(1-p)}{E^2}
$$

Where:

- $n$ = required sample size
- $Z$ = z-value for the chosen confidence level
- $p$ = estimated population proportion
- $1-p$ = estimated proportion without the characteristic
- $E$ = desired margin of error

For approximately 95% confidence, let $Z=1.96$, $p=0.5$, and $E=0.05$:

$$
n=\frac{(1.96)^2(0.5)(1-0.5)}{(0.05)^2}\approx384.16
$$

Round up to 385 observations.

This common formula may need adjustment for finite populations or complex sampling designs.

### Python Implementation

```python
import math

Z = 1.96
p = 0.5
E = 0.05

n = (Z**2 * p * (1 - p)) / (E**2)

print("Calculated Sample Size:", n)
print("Required Sample Size:", math.ceil(n))
```

### AI/ML Example

Sample size affects how much information a machine learning model has for learning and evaluation.

The code calculates a sample size for estimating a population proportion. In AI/ML there is no single universal required sample size; it depends on the task, model complexity, class balance, noise, feature quality, and desired reliability.
