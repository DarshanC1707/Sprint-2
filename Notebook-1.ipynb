## Introduction to Statistics

### 1. What is Statistics?

Definition:
Statistics is the process of collecting, analyzing, and understanding data.

Real-life example:
A shop has daily sales:

₹1000, ₹1500, ₹2000

We can calculate the average sales to understand how much the shop usually earns per day.

```python
import numpy as np

sales = [1000, 1500, 2000]

print(np.mean(sales))
```

```python
The average daily sale is ₹1500.

AI/ML use: Statistics helps us understand data before training a model.
```

### 2. Why Statistics is Important in AI/ML

Definition:
Statistics helps AI/ML models understand data and identify patterns.

Example:
Suppose customer ages are:

20, 25, 30, 35, 40

Finding the average age helps us understand the customers.

```python
ages = [20, 25, 30, 35, 40]

print(np.mean(ages))
```

AI/ML use: Statistics is used in data analysis, preprocessing, feature selection, and model evaluation.

### 3. Statistics vs Machine Learning



Definition:

Statistics helps us understand and analyze data.

Machine Learning uses data to learn patterns and make predictions.

Real-life example – House Price

Real-life Example – House Price:

Suppose we have house sizes and prices.

Statistics: Understand whether larger houses generally have higher prices.

Machine Learning: Predict the price of a new house based on its size and other features.

```python
size = [800, 1000, 1200, 1400]
price = [40, 50, 60, 70]

correlation = np.corrcoef(size, price)

print(correlation)
```

Explanation: The code checks the relationship between house size and price.

AI/ML Use: Statistical analysis helps us understand relationships before building predictive models.

### 4. Descriptive Statistics

Definition:

Descriptive statistics is used to summarize and describe the data we already have.

Examples include mean, median, minimum, maximum, and standard deviation.

Real-life Example:

A company has employee salaries:

₹30,000, ₹35,000, ₹40,000, ₹45,000, ₹50,000

We can summarize them using the average, minimum, and maximum salary.

```python
salary = [30000, 35000, 40000, 45000, 50000]

print("Mean:", np.mean(salary))
print("Minimum:", np.min(salary))
print("Maximum:", np.max(salary))
```

### 5. Inferential Statistics

Definition:

Inferential statistics uses a sample to make conclusions or estimates about a larger population.

Real-life Example:

A company has 10,000 customers.

Instead of asking all customers about their satisfaction, it surveys 500 customers.

The results from those 500 customers are used to understand the larger customer population.

```python
sample_rating = [7, 8, 9, 8, 7]

print("Sample Average:", np.mean(sample_rating))
```

Explanation: The sample average can be used as an estimate of the population's average satisfaction.

### 6. Population

Definition:

Population is the entire group that we want to study.

Real-life Example:

A college has 2,000 students.

If we want to study the height of students:

All 2,000 students = Population

```python
population = [165, 170, 168, 175, 172]

print("Population Size:", len(population))
```

Explanation: The complete list represents our small example population.

### 7. Sample

Definition:

A sample is a smaller group selected from a population.

Real-life Example:

A college has 2,000 students.

We select 100 students to study their average height.

100 students = Sample

```python
population = [160, 165, 170, 175, 180]

sample = population[:3]

print("Sample:", sample)
```

Explanation: Three values were selected from the larger population.

### 8. Census

Definition:

A census means collecting information from every member of a population.

Real-life Example:

A company has 100 employees.

HR collects salary information from all 100 employees.

This is a census because everyone is included.

```python
employees = [1, 2, 3, 4, 5]

print("Employees Studied:", len(employees))
```

Explanation: Every member of our example population is included.

### 9. Sampling

Definition:

Sampling is the process of selecting a smaller sample from a larger population.

Real-life Example:

A factory produces 10,000 bottles.

Quality control randomly checks 100 bottles.

Selecting those 100 bottles is called sampling.

```python
import random

bottles = list(range(1, 101))

sample = random.sample(bottles, 10)

print(sample)
```

Explanation: The program randomly selects 10 bottles from the population.

### 11. Features

Definition:

Features are the input variables given to a machine-learning model.

Real-life Example:

To predict a house price:

House size → Feature

Bedrooms → Feature

House age → Feature

The model uses the features to predict the target.

### 12. Observation

Definition:

An observation is one individual record or row in a dataset.

Real-life Example:


Name---- Age ---- Marks

Ravi --- 20	 ----  85

Anu	 --- 21	 ---- 90

Ravi, 20, 85 represents one observation.


```python
student = {
    "Name": "Ravi",
    "Age": 20,
    "Marks": 85
}

print(student)
```

Explanation: The dictionary represents one student's complete observation.

### 13. Dataset

Definition:

A dataset is a collection of related observations and variables.

Real-life Example:

A hospital stores:

Age -- Weight ---  BP

25 	--  60    --- 120

35  --  70	  ---  125

45	--  80	  ---  135

The complete table is a dataset.

```python
data = pd.DataFrame({
    "Age": [25, 35, 45],
    "Weight": [60, 70, 80],
    "BP": [120, 125, 135]
})

print(data)
```

Explanation: The DataFrame contains multiple observations and variables.

AI/ML Use: Datasets are used to train, validate, and test ML models.

### 14. Quantitative Data


Definition:

Quantitative data is numerical data that can be measured or counted.




Real-life Example:

A person's:

Age = 25

Height = 170 cm

Weight = 70 kg

These are quantitative because they contain numerical values.

```python
ages = [20, 25, 30]

print("Average Age:", np.mean(ages))
```

Explanation: Mathematical calculations can be performed on quantitative data.

AI/ML Use: Numerical values are commonly used as input features in ML models.

### 15. Qualitative Data

Definition:

Qualitative data represents categories, names, or labels instead of numerical measurements.

Real-life Example:

For a customer:

City = Bengaluru

Payment = UPI

Membership = Premium

These represent categories rather than measured quantities.

```python
payments = ["UPI", "Card", "UPI", "Cash"]

print(pd.Series(payments).value_counts())
```

Explanation: The code counts how many customers used each payment type.

AI/ML Use: Qualitative data is often encoded into numerical values before model training.

### 16. Continuous Data

Definition:

Continuous data can take any value within a range, including decimal values.

Real-life Example:
    
A person's weight could be:

70 kg

70.5 kg

70.55 kg

Therefore, weight is continuous data.

```python
weights = [70.5, 68.2, 75.7]

print("Average Weight:", np.mean(weights))
```

Explanation: Weight can contain decimal measurements.

AI/ML Use: Continuous features are common in regression and many other ML models.

### 17. Discrete Data

Definition:

Discrete data contains separate and countable values, usually whole numbers.

Real-life Example:

Number of children in a family:

1, 2, 3, 4

A family can have 2 children, but not 2.5 children.

Therefore, this is discrete data.

```python
children = [2, 1, 3, 2]

print("Total:", sum(children))
```

Explanation: The values represent countable quantities.

AI/ML Use: Used for count-based features such as number of purchases, calls, clicks, or visits.

### 18. Structured Data

Definition:

Structured data is data stored in a fixed and organized format, usually rows and columns.

Real-life Example:

A bank stores customer information:

| Name | Age | Balance |
|------|-----|---------|
| Ravi | 25  | 30000   |
| Anu  | 28  | 45000   |

Every value has a defined row and column, so it is structured data.

```python
customers = pd.DataFrame({
    "Name": ["Ravi", "Anu"],
    "Age": [25, 28],
    "Balance": [30000, 45000]
})

print(customers)
```

### 19. Unstructured Data

Definition:

Unstructured data is data that does not naturally follow a fixed row-and-column format.

Real-life Example:
    
A customer writes:

"The phone camera is excellent, but the battery could be better."

This is free-form text.

Other examples are images, videos, audio recordings, emails, and documents.

```python
review = "The phone camera is excellent"

words = review.split()

print(words)
```

Explanation: The program converts the free-form sentence into individual words.

AI/ML Use: Unstructured data is heavily used in NLP, computer vision, speech recognition, and LLMs.
