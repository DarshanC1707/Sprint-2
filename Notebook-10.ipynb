#  Hypothesis Testing



# 1. Null Hypothesis

### Definition

The null hypothesis ($H_0$) is the default statement that there is no significant difference, effect, or relationship.

### Real-life Example

A company claims average battery life is 10 hours, so $H_0: \mu=10$.

### Formula / Numerical Example

$$H_0:\mu=\mu_0$$
Where:
- $H_0$ = null hypothesis
- $\mu$ = population mean
- $\mu_0$ = claimed population mean

Example: $H_0:\mu=10$ means we initially assume the population average is 10 hours.

### Python Implementation


```python
claimed_mean=10
sample_mean=10.4
print("Claimed Mean:",claimed_mean)
print("Sample Mean:",sample_mean)
print("H0: Population mean = 10")
```

### AI/ML Example

In AI/ML, the null hypothesis can state that two models have the same average performance. We begin with no meaningful difference and use statistical evidence to decide whether to reject that assumption.


# 2. Alternative Hypothesis

### Definition

The alternative hypothesis ($H_1$ or $H_a$) states that a significant difference, effect, or relationship exists.

### Real-life Example

If the null says average battery life is 10 hours, a two-sided alternative is $H_1:\mu\neq10$.

### Formula / Numerical Example

Two-sided: $$H_1:\mu\neq\mu_0$$
One-sided: $$H_1:\mu>\mu_0$$ or $$H_1:\mu<\mu_0$$
Where $H_1$ = alternative hypothesis, $\mu$ = population mean, and $\mu_0$ = claimed mean.

### Python Implementation


```python
print("H0: Population mean = 10")
print("H1: Population mean != 10")
```

### AI/ML Example

In AI/ML, the alternative hypothesis may state that two models have different average performances. It represents the effect or difference for which we are seeking evidence.


# 3. p-value

### Definition

The p-value measures how compatible the observed result, or a more extreme result, is with the null hypothesis under the assumptions of the test.

### Real-life Example

If $p=0.02$ and $\alpha=0.05$, we reject $H_0$ because $0.02<0.05$.

### Formula / Numerical Example

$$p\leq\alpha\Rightarrow\text{Reject }H_0$$
$$p>\alpha\Rightarrow\text{Fail to reject }H_0$$
Where $p$ = p-value and $\alpha$ = significance level.

### Python Implementation


```python
from scipy.stats import ttest_1samp
scores=[72,75,71,74,73,76,72,75]
t_stat,p_value=ttest_1samp(scores,popmean=70)
print("p-value:",p_value)
print("Reject H0" if p_value<0.05 else "Fail to reject H0")
```

### AI/ML Example

In AI/ML, p-values can be used when statistically comparing model results. A small p-value gives evidence against the null hypothesis, but it does not tell us the size or practical importance of a difference.


# 4. Confidence Interval

### Definition

A confidence interval gives a range of plausible values for an unknown population parameter using sample data.

### Real-life Example

Instead of estimating average delivery time as exactly 30 minutes, we may report a 95% confidence interval such as 28–32 minutes.

### Formula / Numerical Example

For a mean with known population standard deviation:
$$CI=\bar{x}\pm z_{\alpha/2}\frac{\sigma}{\sqrt{n}}$$
Where:
- $CI$ = confidence interval
- $\bar{x}$ = sample mean
- $z_{\alpha/2}$ = critical Z value
- $\sigma$ = population standard deviation
- $n$ = sample size

If $\bar{x}=50$, $\sigma=10$, $n=100$, and $z=1.96$:
$$CI=50\pm1.96=(48.04,51.96)$$

### Python Implementation


```python
import numpy as np
from scipy.stats import norm
mean,sigma,n=50,10,100
z=norm.ppf(0.975)
margin=z*sigma/np.sqrt(n)
print("95% CI:",(mean-margin,mean+margin))
```

### AI/ML Example

In AI/ML, confidence intervals can show uncertainty around estimated model performance. They provide more information than reporting only one estimated score.


# 5. Significance Level

### Definition

The significance level is a threshold chosen before testing for deciding when evidence against the null hypothesis is strong enough to reject it.

### Real-life Example

A common choice is $\alpha=0.05$. If the p-value is at or below 0.05, we reject $H_0$ using that rule.

### Formula / Numerical Example

$$\alpha=0.05$$
Decision rule: $$p\leq\alpha$$
Where $\alpha$ = alpha or significance level and $p$ = p-value.

### Python Implementation


```python
alpha=0.05
p_value=0.03
print("Reject H0" if p_value<=alpha else "Fail to reject H0")
```

### AI/ML Example

In AI/ML experiments, a significance level can be chosen before comparing models. It controls the probability of a Type I error under the null hypothesis.


# 6. Type I Error

### Definition

A Type I error occurs when we reject the null hypothesis even though it is actually true. It is similar to a false positive.

### Real-life Example

A medical test incorrectly says a healthy person has a disease.

### Formula / Numerical Example

$$P(\text{Type I Error})=\alpha$$
Where $P$ = probability and $\alpha$ = significance level.
For $\alpha=0.05$, the Type I error probability is 5% when $H_0$ is true.

### Python Implementation


```python
alpha=0.05
print("Type I Error Rate:",alpha*100,"%")
```

### AI/ML Example

In a fraud-detection example, a false positive means incorrectly flagging a legitimate transaction as fraudulent. This is analogous to a Type I error.


# 7. Type II Error

### Definition

A Type II error occurs when we fail to reject the null hypothesis even though it is false. It is similar to a false negative.

### Real-life Example

A medical test fails to detect a disease that is actually present.

### Formula / Numerical Example

$$P(\text{Type II Error})=\beta$$
$$Power=1-\beta$$
Where $\beta$ = beta, the probability of Type II error.
If $\beta=0.20$, then $Power=0.80$.

### Python Implementation


```python
beta=0.20
power=1-beta
print("Type II Error Rate:",beta)
print("Statistical Power:",power)
```

### AI/ML Example

In fraud detection, a false negative means failing to identify a fraudulent transaction. This is analogous to a Type II error.


# 8. One-tailed Test

### Definition

A one-tailed test checks for an effect in one specified direction.

### Real-life Example

A company tests whether a new training method increases the average score above 70: $H_1:\mu>70$.

### Formula / Numerical Example

For a Z-test:
$$z=\frac{\bar{x}-\mu_0}{\sigma/\sqrt{n}}$$
Where:
- $z$ = Z test statistic
- $\bar{x}$ = sample mean
- $\mu_0$ = hypothesized mean
- $\sigma$ = population standard deviation
- $n$ = sample size

### Python Implementation


```python
from scipy.stats import ttest_1samp
scores=[72,74,75,73,76,77,74,75]
t_stat,p_two=ttest_1samp(scores,popmean=70)
p_one=p_two/2 if t_stat>0 else 1-p_two/2
print("One-tailed p-value:",p_one)
```

### AI/ML Example

In AI/ML, a one-tailed test can be used when the predefined question is specifically whether a new model improves a metric. The direction should be chosen before examining the results.


# 9. Two-tailed Test

### Definition

A two-tailed test checks whether a parameter differs from a specified value in either direction.

### Real-life Example

We test whether average exam score is different from 70: $H_0:\mu=70$ and $H_1:\mu\neq70$.

### Formula / Numerical Example

For a Z-test:
$$z=\frac{\bar{x}-\mu_0}{\sigma/\sqrt{n}}$$
If $\alpha=0.05$, each tail receives:
$$\frac{\alpha}{2}=0.025$$
Where $\alpha$ = significance level, $\bar{x}$ = sample mean, $\mu_0$ = hypothesized mean, $\sigma$ = population standard deviation, and $n$ = sample size.

### Python Implementation


```python
from scipy.stats import ttest_1samp
scores=[65,68,72,74,69,71,73,67]
t_stat,p_value=ttest_1samp(scores,popmean=70)
print("Two-tailed p-value:",p_value)
print("Reject H0" if p_value<0.05 else "Fail to reject H0")
```

### AI/ML Example

In AI/ML, a two-tailed test can check whether a new model's average performance differs from a baseline without specifying whether it will be higher or lower.
