#  Measures of Dispersion



# 1. Range

### Definition

Definition

Range is the difference between the maximum value and minimum value in a dataset.

### Real-life Example

Suppose the temperatures recorded during five days are:

20°C, 25°C, 30°C, 32°C, 35°C

The difference between the highest and lowest temperature gives the range.

### Formula

$$
\text{Range} = \text{Maximum Value} - \text{Minimum Value}
$$

### Numerical Example

Data: 10, 20, 30, 40, 50

$$
\text{Range} = 50 - 10 = 40
$$

### Python Implementation

```python
data = [10, 20, 30, 40, 50]
data_range = max(data) - min(data)
print("Range:", data_range)
```

### AI/ML Example

Range can be used to understand the spread of feature values before preprocessing.

```python
ages = [20, 25, 30, 35, 60]
age_range = max(ages) - min(ages)
print("Age Range:", age_range)
```

### Explanation

This code finds the difference between the highest and lowest age.

A large range tells us that the feature values are spread across a wider interval. Checking the range can help us understand the scale of features and notice unusually extreme values before training a model.

# 2. Variance

### Definition

Variance measures how far the values in a dataset are spread from their mean.

A higher variance means the values are more spread out, while a lower variance means they are closer to the mean.

### Real-life Example

Student A: 78, 80, 82  
Student B: 50, 80, 110

Both have the same mean, but Student B's marks vary much more.

### Formula


### Population Variance

$$
\sigma^2 = \frac{\sum (x_i-\mu)^2}{N}
$$

- **$x_i$ (X sub i)** → Each individual value in the dataset
- **$\mu$ (Mu)** → Population mean (average)

### Sample Variance

$$
s^2 = \frac{\sum (x_i-\bar{x})^2}{n-1}
$$

- **$x_i$ (X sub i)** → Each individual value in the dataset
- **$\bar{x}$ (X bar)** → Sample mean (average)

### Numerical Example

Data: 2, 4, 6

Here,

$$
x_1 = 2,\quad x_2 = 4,\quad x_3 = 6
$$

So, **$x_i$ represents each individual value: 2, 4, and 6.**

Calculate the mean:

$$
\bar{x} = \frac{2+4+6}{3}=4
$$

So, **$\bar{x}=4$ is the sample mean.**

If the data 2, 4, 6 represents the entire population:

$$
\mu = 4
$$

So, **$\mu=4$ is the population mean.**

### Population Variance Calculation

$$
\sigma^2 = \frac{(2-4)^2+(4-4)^2+(6-4)^2}{3}
$$

$$
= \frac{4+0+4}{3}
$$

$$
= \frac{8}{3} \approx 2.67
$$


### Python Implementation

```python
import numpy as np
data = [2, 4, 6]
variance = np.var(data)
print("Variance:", variance)
```

### AI/ML Example

Variance can help identify features that contain very little variation.

```python
import numpy as np
feature = [10, 10, 10, 10, 11]
variance = np.var(feature)
print("Feature Variance:", variance)
```

### Explanation

This code measures how much the feature values vary from their mean.

If a feature has almost zero variance, most values are nearly identical. Such a feature may provide very little useful information to a machine learning model.

Variance is therefore used in techniques such as variance-based feature selection.

# 3. Standard Deviation

### Definition

Standard deviation measures how far values typically spread from the mean.

It is the square root of variance.

### Real-life Example

If students in one class score close to the average, the class has a low standard deviation. If the marks are widely spread, it has a high standard deviation.

### Formula

$$
\sigma = \sqrt{\frac{\sum(x_i-\mu)^2}{N}}
$$

### Numerical Example

$$
\sigma^2 = 2.67
$$

$$
\sigma = \sqrt{2.67} \approx 1.63
$$

### Python Implementation

```python
import numpy as np
data = [2, 4, 6]
std_dev = np.std(data)
print("Standard Deviation:", std_dev)
```

### AI/ML Example

Standard deviation is used in StandardScaler to standardize numerical features.

```python
from sklearn.preprocessing import StandardScaler

data = [[20], [30], [40], [50], [60]]
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data)
print(scaled_data)
```

### Explanation

StandardScaler calculates the mean and standard deviation of the feature.

It transforms each value using:

$$
z = \frac{x-\mu}{\sigma}
$$

After scaling, the feature has approximately mean 0 and standard deviation 1.

This helps algorithms such as KNN, SVM and Logistic Regression when features have different scales.

# 4. Quartiles

### Definition

Quartiles divide an ordered dataset into four equal parts.

Q1 = 25% of the data  
Q2 = 50% of the data and is the median  
Q3 = 75% of the data

### Real-life Example

Students arranged according to exam marks can be divided into four groups using quartiles.

### Formula

$$
Q_1 = 25^{th}\text{ percentile}
$$

$$
Q_2 = 50^{th}\text{ percentile}
$$

$$
Q_3 = 75^{th}\text{ percentile}
$$

### Numerical Example

Data: 10, 20, 30, 40, 50, 60, 70

$$
Q_1 = 20,\quad Q_2 = 40,\quad Q_3 = 60
$$

### Python Implementation

```python
import numpy as np
data = [10, 20, 30, 40, 50, 60, 70]
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)
print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

### AI/ML Example

Quartiles are commonly used while analysing the distribution of numerical features.

```python
import pandas as pd

df = pd.DataFrame({"Salary": [20000,25000,30000,35000,40000,45000,50000]})
print(df["Salary"].quantile([0.25, 0.50, 0.75]))
```

### Explanation

The code divides the salary distribution into quartile positions.

Q1 represents the 25th percentile, Q2 represents the median, and Q3 represents the 75th percentile.

Quartiles help us understand feature distribution and are also used to calculate IQR for detecting possible outliers.

# 5. Percentiles

### Definition

Percentiles divide an ordered dataset into 100 parts and show the position of a value relative to the rest of the data.

### Real-life Example

If a student is in the 90th percentile, the student's score is at or above roughly 90% of the scores, depending on the percentile convention.

### Formula

$$
\text{Percentile Rank} =
\frac{\text{Number of values below }x}{\text{Total number of values}}\times100
$$

### Numerical Example

$$
\text{Percentile Rank} = \frac{80}{100}\times100 = 80
$$

### Python Implementation

```python
import numpy as np
data = [10, 20, 30, 40, 50]
p80 = np.percentile(data, 80)
print("80th Percentile:", p80)
```

### AI/ML Example

Percentiles can be used to identify unusually large feature values.

```python
import numpy as np
values = [10, 20, 30, 40, 50, 60, 100]
threshold = np.percentile(values, 95)
print("95th Percentile:", threshold)
```

### Explanation

The code calculates the value corresponding to the 95th percentile.

Values above a chosen high percentile can be inspected as potentially unusual observations.

Percentile-based thresholds are useful during data preprocessing because they do not depend on the data following a normal distribution.

# 6. Interquartile Range (IQR)

### Definition

Interquartile Range measures the spread of the middle 50% of the dataset.

It is calculated as the difference between Q3 and Q1.

### Real-life Example

If most employee salaries are between ₹30,000 and ₹60,000 while a few managers earn much more, IQR focuses on the middle portion and is less affected by extreme values.

### Formula

$$
IQR = Q_3-Q_1
$$

### Numerical Example

If Q1 = 20 and Q3 = 60:

$$
IQR = 60-20=40
$$

$$
\text{Lower Bound} = Q_1 - 1.5(IQR)
$$

$$
\text{Upper Bound} = Q_3 + 1.5(IQR)
$$

### Python Implementation

```python
import numpy as np
data = [10,20,30,40,50,60,70]
q1 = np.percentile(data,25)
q3 = np.percentile(data,75)
iqr = q3-q1
print("IQR:", iqr)
```

### AI/ML Example

IQR is commonly used to detect possible outliers before model training.

```python
import numpy as np
data = np.array([10,12,14,15,16,18,20,100])
q1 = np.percentile(data,25)
q3 = np.percentile(data,75)
iqr = q3-q1
lower_bound = q1 - 1.5*iqr
upper_bound = q3 + 1.5*iqr
outliers = data[(data < lower_bound) | (data > upper_bound)]
print("Outliers:", outliers)
```

### Explanation

The code calculates Q1, Q3 and IQR and creates lower and upper boundaries using 1.5 × IQR.

Values outside these boundaries are flagged as potential outliers.

Here, 100 is identified as an outlier. Detecting extreme observations can be important because they can strongly affect some machine learning models.

# 7. Coefficient of Variation

### Definition

Coefficient of Variation, or CV, measures the standard deviation relative to the mean.

It allows us to compare variability between datasets with different averages or scales, provided the mean is meaningful and not near zero.

### Real-life Example

For two investments with different average returns, CV compares the variation relative to each investment's average return.

### Formula

$$
CV = \frac{\text{Standard Deviation}}{\text{Mean}}\times100
$$

### Numerical Example

Mean = 50 and Standard Deviation = 10

$$
CV = \frac{10}{50}\times100 =20\%
$$

### Python Implementation

```python
import numpy as np
data = [40,45,50,55,60]
mean = np.mean(data)
std_dev = np.std(data)
cv = (std_dev/mean)*100
print("Coefficient of Variation:", cv, "%")
```

### AI/ML Example

CV can be used to compare the relative variability of numerical features that have different means.

```python
import numpy as np
feature1 = [100,110,120,130,140]
feature2 = [10,15,20,25,30]
cv1 = (np.std(feature1)/np.mean(feature1))*100
cv2 = (np.std(feature2)/np.mean(feature2))*100
print("Feature 1 CV:", cv1)
print("Feature 2 CV:", cv2)
```

### Explanation

The code calculates the standard deviation relative to the mean for each feature.

Because CV is expressed as a percentage, it can compare relative variability when features have different average values.

A larger CV indicates greater variability relative to the mean. CV should be used carefully when the mean is zero or very close to zero.
